```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

nama_gambar = 'bro me.jpeg'  # change if needed

# read image
image = cv2.imread(nama_gambar)

if image is None:
    print(f"Error: Gambar '{nama_gambar}' tidak ditemukan!")
else:
    # convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # set m-bit (example: 4-bit)
    m = 4
    levels = 2 ** m

    # quantization (reduce levels)
    mbit = np.floor(gray / (256 / levels)) * (256 / levels)
    mbit = mbit.astype('uint8')

    # show images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(gray, cmap='gray')
    plt.title('Grayscale')
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(mbit, cmap='gray')
    plt.title(f'{m}-bit Image')
    plt.axis('off')

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

nama_gambar = 'bro me.jpeg'  # change if needed

# read image
image = cv2.imread(nama_gambar)

if image is None:
    print(f"Error: Gambar '{nama_gambar}' tidak ditemukan!")
else:
    # convert BGR to RGB (for display)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # increase brightness
    bright = cv2.convertScaleAbs(image, alpha=1, beta=50)
    bright_rgb = cv2.cvtColor(bright, cv2.COLOR_BGR2RGB)

    # show images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(image_rgb)
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(bright_rgb)
    plt.title('Bright Image')
    plt.axis('off')

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

nama_gambar = 'bro me.jpeg'  # change if needed

# read image
image = cv2.imread(nama_gambar)

if image is None:
    print(f"Error: Gambar '{nama_gambar}' tidak ditemukan!")
else:
    # convert BGR to RGB (for display)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # adjust contrast
    contrast = cv2.convertScaleAbs(image, alpha=1.5, beta=0)
    contrast_rgb = cv2.cvtColor(contrast, cv2.COLOR_BGR2RGB)

    # show images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(image_rgb)
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(contrast_rgb)
    plt.title('High Contrast')
    plt.axis('off')

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

nama_gambar = 'bro me.jpeg'  # change if needed

# read image
image = cv2.imread(nama_gambar)

if image is None:
    print(f"Error: Gambar '{nama_gambar}' tidak ditemukan!")
else:
    # convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # show grayscale image
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(gray, cmap='gray')
    plt.title('Grayscale Image')
    plt.axis('off')

    # show histogram
    plt.subplot(1,2,2)
    plt.hist(gray.ravel(), bins=256, range=[0,256])
    plt.title('Histogram')

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

nama_gambar = 'bro me.jpeg'  # change if needed

# read image
image = cv2.imread(nama_gambar)

if image is None:
    print(f"Error: Gambar '{nama_gambar}' tidak ditemukan!")
else:
    # convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # apply histogram equalization
    equalized = cv2.equalizeHist(gray)

    # show images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(gray, cmap='gray')
    plt.title('Grayscale')
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(equalized, cmap='gray')
    plt.title('Histogram Equalization')
    plt.axis('off')

    plt.show()
```
